```python
from indicnlp.tokenize import indic_tokenize


# Telugu paragraph used for the homework.
paragraph = (
    "హైదరాబాద్‌లో ఈ రోజు వాతావరణం చాలా ఆహ్లాదకరంగా ఉంది. "
    "ప్రజలు ఉదయం త్వరగా లేచి పార్కులో నడవడానికి వెళ్తారు. "
    "పిల్లలు కలిసి ఆటలు ఆడుతూ కొత్త కథలు చెప్పుకున్నారు. "
    "సాయంత్రం మేము మా స్నేహితులతో కలిసి టీ తాగడానికి వెళ్లాము."
)


# ---------------------------------------------------------
# 1. Naive space-based tokenization
# ---------------------------------------------------------

space_tokens = paragraph.split()

print("Naive space-based tokens:")
print(space_tokens)


# ---------------------------------------------------------
# 2. Manual tokenization
# ---------------------------------------------------------

manual_tokens = [
    "హైదరాబాద్‌లో", "ఈ", "రోజు", "వాతావరణం", "చాలా",
    "ఆహ్లాదకరంగా", "ఉంది", ".", "ప్రజలు", "ఉదయం",
    "త్వరగా", "లేచి", "పార్కులో", "నడవడానికి", "వెళ్తారు", ".",
    "పిల్లలు", "కలిసి", "ఆటలు", "ఆడుతూ", "కొత్త",
    "కథలు", "చెప్పుకున్నారు", ".", "సాయంత్రం", "మేము",
    "మా", "స్నేహితులతో", "కలిసి", "టీ", "తాగడానికి",
    "వెళ్లాము", "."
]

print("\nManual tokens:")
print(manual_tokens)


# ---------------------------------------------------------
# 3. Indic NLP tokenizer
# ---------------------------------------------------------

tool_tokens = indic_tokenize.trivial_tokenize(paragraph)

print("\nIndic NLP tool tokens:")
print(tool_tokens)


# ---------------------------------------------------------
# 4. Compare manual tokens with the NLP tool
# ---------------------------------------------------------

print("\nTokens from the tool that differ from manual tokens:")

for index, (manual, tool) in enumerate(
    zip(manual_tokens, tool_tokens)
):
    if manual != tool:
        print(
            f"Position {index}: "
            f"manual='{manual}', tool='{tool}'"
        )
```
